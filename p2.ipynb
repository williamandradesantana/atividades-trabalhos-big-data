# Problema 2: Produtividade de Desenvolvedores

Conjunto de Dados: Uma equipe de desenvolvimento registrou o número de tarefas 
concluídas por semana e as horas de treinamento realizadas por um grupo de 10 
desenvolvedores.

## Análise Estatística Solicitada:
1. Calcule a média, mediana, quartis e desvio padrão para as variáveis Tarefas 
Concluídas e Horas de Treinamento.
2. Calcule a covariância e o coeficiente de correlação (r) entre as duas variáveis.

## Inferência 
- Qual a relação entre o número de horas de treinamento e a produtividade dos desenvolvedores? Os dados indicam que mais treinamento está associado a mais tarefas concluídas?
- **Resposta**: Neste conjunto de dados, quanto maior o número de horas de treinamento, maior tende a ser o número de tarefas concluídas. Visto que r = 0,9735, há uma relação positiva muito forte entre os conjuntos. A covariância também é positiva: Cov(X,Y) = 8,3889, isso reforça que as duas variáveis tendem a crescer juntas.


```python
data_set = {"tarefas_concluidas": [10, 12, 8, 15, 11, 9, 14, 16, 7, 13],
            "horas_treinamento": [5, 8, 4, 10, 6, 5, 9, 12, 3, 7]}
```

```python
# Imports
import numpy as np
import matplotlib.pyplot as plt
```

```python
'''Cálculo de estatísticas descritivas'''

# Média
media_tarefas_concluidas = np.mean(data_set["tarefas_concluidas"])
media_horas_treinamento = np.mean(data_set["horas_treinamento"])

# Mediana
mediana_tarefas_concluidas = np.median(data_set["tarefas_concluidas"])
mediana_horas_treinamento = np.median(data_set["horas_treinamento"])

# Quartis
quartis_tarefas_concluidas = np.percentile(data_set["tarefas_concluidas"], [25, 50, 75])
quartis_horas_treinamento = np.percentile(data_set["horas_treinamento"], [25, 50, 75]) 

# Desvio padrão
desvio_padrao_tarefas_concluidas = np.std(data_set["tarefas_concluidas"])
desvio_padrao_horas_treinamento = np.std(data_set["horas_treinamento"])

# Covariância e Coeficiente de Correlação (r)
covariancia = np.cov(data_set["tarefas_concluidas"], data_set["horas_treinamento"])[0][1]
coeficiente_correlacao = np.corrcoef(data_set["tarefas_concluidas"], data_set["horas_treinamento"])[0][1]
```

```python
'''Plot e Inferência'''

print("Média de tarefas concluídas:", media_tarefas_concluidas)
print("Média de horas de treinamento:", media_horas_treinamento)
print("Mediana de tarefas concluídas:", mediana_tarefas_concluidas)
print("Mediana de horas de treinamento:", mediana_horas_treinamento)
print("Quartis de tarefas concluídas:", quartis_tarefas_concluidas)
print("Quartis de horas de treinamento:", quartis_horas_treinamento)
print("Desvio padrão de tarefas concluídas:", desvio_padrao_tarefas_concluidas)
print("Desvio padrão de horas de treinamento:", desvio_padrao_horas_treinamento)
print("Covariância entre tarefas concluídas e horas de treinamento:", covariancia)
print("Coeficiente de Correlação (r) entre tarefas concluídas e horas de treinamento:", coeficiente_correlacao)

plt.boxplot([
    data_set["tarefas_concluidas"],
    data_set["horas_treinamento"]
])
plt.xticks(
    [1, 2],
    ["Tarefas concluídas", "Horas de treinamento"]
)
plt.ylabel("Valor")
plt.title("Distribuição das Variáveis")
plt.show()

x = np.array(data_set["horas_treinamento"])
y = np.array(data_set["tarefas_concluidas"])

# Regressão linear
coef = np.polyfit(x, y, 1)
reta = np.poly1d(coef)

# Plot
plt.scatter(x, y, label="Dados")
plt.plot(x, reta(x), label="Regressão linear")
plt.title("Relação entre Horas de Treinamento e Tarefas Concluídas")
plt.xlabel("Horas de Treinamento")
plt.ylabel("Tarefas Concluídas")
plt.legend()
plt.show()
```
