#### Calcule a média, mediana, quartis e desvio padrão para as variáveis Dados Limpos e Tempo de Processamento.

```python
import pandas as pd

dados = {
    "dados_limpos_percentual": [95, 98, 85, 92, 80, 90, 99, 88, 82, 96, 94, 87, 84],
    "tempo_processamento_segundos": [12, 10, 20, 15, 24, 18, 9, 21, 23, 11, 14, 19, 22],
}

df = pd.DataFrame(dados)

media = df.mean()
mediana = df.median()
quartis = df.quantile([0.25, 0.5, 0.75])
desvio_padrao = df.std()


def print_result(title_result, result):
    print(f"\n\n----{title_result.upper()}----")
    print(media)


print_result("Média", media)
print_result("Mediana", mediana)
print_result("Quartis", quartis)
print_result("Desvio Padrão", desvio_padrao)
```

#### Calcule a covariância e o coeficiente de correlação (r) entre as duas variáveis.

```python
covariancia = df["dados_limpos_percentual"].cov(df["tempo_processamento_segundos"])
correlacao = df["dados_limpos_percentual"].corr(df["tempo_processamento_segundos"])

print("Covariância: ", covariancia)
print("Correlação: ", correlacao)
```

#### Como a qualidade dos dados afeta o tempo de processamento de um algoritmo? Uma alta porcentagem de dados limpos se traduz em um tempo de processamento mais rápido?
A qualidade dos dados pode afetar o tempo de processamento de um algoritmo, pois dados mais limpos e organizados podem reduzir a quantidade de operações necessárias durante o processamento. De acordo com os dados apresentados, existe uma forte correlação negativa entre a porcentagem de dados limpos e o tempo de processamento (r ≈ -0,982). Isso indica que, neste conjunto de dados, uma maior porcentagem de dados limpos está associada a um menor tempo de processamento.
